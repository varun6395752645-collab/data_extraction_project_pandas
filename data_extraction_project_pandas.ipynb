```python
#Function for extracting the episodes:
def extract_episodes(text):
    check=False
    data=""
    for i in text:
        if i==")":
            check=False
            break
        if check==True:
            data+=i
        if i=="(":
            check=True

    return data
```

```python
df["episodes"]=df["Title"].apply(extract_episodes)
```

```python
df["episodes"]=df["episodes"].str.replace(" eps","")
```

```python
#function for extracting time stamp:
def extract_timestamp(text):
    check=False
    data=""

    for i in range(len(text)):
        if text[i]==",":
            check=False
            break
            
        if text[i] == ")":
            check=True
            if check==True:
                j=i+1
                while j<i+20:
                    data+=text[j]
                    j+=1
    return data
```

```python
df["Time_Stamp"]=df["Title"].apply(extract_timestamp)
```
